```python
from matplotlib.axes import Axes
import matplotlib.pyplot as plt

# 展示中文
plt.rcParams['font.sans-serif'] = ['SimHei'] # 设置中文字体为黑体

figure, axes = plt.subplots(nrows=1, ncols=2, figsize=(20,5), dpi=100) # nrows表示行数，ncols表示列数，figsize表示图形大小，dpi表示分辨率

# 绘制柱状图
countries = ['委内瑞拉', '沙特阿拉伯', '加拿大', '伊朗', '伊拉克', '科威特', '阿联酋', '俄罗斯', '利比亚', '尼日利亚']
reserves = [303.2, 297.6, 168.9, 157.8, 145.0, 101.5, 97.8, 80.0, 48.4, 37.1]
axes1:Axes = axes[0] # 获取第一个子图
axes1.bar(countries, reserves, width=0.5, color='orange') # 绘制柱状图
axes1.set_title("世界石油储量前十国家", fontsize=18) # 设置标题
axes1.set_xlabel("国家", fontsize=14) # 设置x轴标签
axes1.set_ylabel("储量(亿桶)", fontsize=14) # 设置y轴标签
axes1.grid(linestyle='--', alpha=0.5) # 显示网格

# 绘制饼图
countries2 = ['委内瑞拉', '沙特阿拉伯', '加拿大', '伊朗', '伊拉克', '科威特', '阿联酋', '俄罗斯', '其他国家']
reserves2 = [303.2, 297.6, 168.9, 157.8, 145.0, 101.5, 97.8, 80.0, 48.4]
axes2:Axes = axes[1] # 获取第二个子图
axes2.pie(reserves2, labels=countries2, autopct='%1.1f%%', startangle=90) #绘制饼图
axes2.set_title("世界石油储量前十国家占比", fontsize=18) # 设置标题
axes2.legend(loc='lower center', ncol=5, bbox_to_anchor=(0.5, -0.2)) # 显示图例, ncol表示图例列数，bbox_to_anchor表示图例位置(0.5: 控制水平位置, -0.2: 控制垂直位置)表示图例在x轴中间，y轴下方

plt.savefig('../data/01.png') # 保存图形为png格式

plt.show() # 显示图形
```
