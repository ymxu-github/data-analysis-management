```python
 import pandas as pd

# 读取 CSV 文件
df = pd.read_csv("../data/sales.csv")
df['销售金额'] = df['单价'] * df['销售数量'] # 计算销售金额
df.groupby('产品名称')['订单号'].count() # 按照产品名称分组，统计每个产品的数量
df.groupby('产品名称')['销售数量'].sum() # 按照产品名称分组，统计每个产品的销售数量总和
df.groupby('产品名称')['销售金额'].sum() # 按照产品名称分组，统计每个产品的销售金额总和
df.groupby('产品名称')['单价'].min() # 按照产品名称分组，统计每个产品的最低单价
df.groupby('产品名称')['单价'].max() # 按照产品名称分组，统计每个产品的最高单价
df.groupby('产品名称')['单价'].mean() # 按照产品名称分组，统计每个产品的平均单价
df.groupby('产品名称')['单价'].agg(['min', 'max', 'mean']) # 按照产品名称分组，统计每个产品的最低单价、最高单价和平均单价
df.groupby('产品名称').agg({'销售数量': ['sum', 'mean'], '销售金额': ['sum', 'mean']}).to_dict() # 按照产品名称分组，统计每个产品的销售数量和销售金额的总和和平均值
```
