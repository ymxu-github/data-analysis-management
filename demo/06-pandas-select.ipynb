```python
import pandas as pd
# 读取 CSV 文件
df = pd.read_csv("../data/sales.csv", usecols=['订单号', '产品类别', '产品名称', '销售数量', '单价'])

# 选择列数据
df['产品名称'] # 选择单列
df[['产品类别', '销售数量']] # 选择多列

# 选择行数据
# iloc 基于行号, 语法：iloc[start:stop:step], 不包含 stop
df.iloc[0:5] # 选择前 5 行数据

# loc 基于索引标签, 语法：loc[start:stop:step], 包含 stop
df.loc[0:5] # 选择前 6 行数据, 包含第 5 行数据
```
