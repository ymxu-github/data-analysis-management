```python
import random
import matplotlib.pyplot as plt

# 展示中文
plt.rcParams['font.sans-serif'] = ['SimHei'] # 设置中文字体为黑体

# 绘制折线图
x = [i for i in range(1, 25)]
y_bj = [random.randint(10, 30) for i in x]
y_xa = [random.randint(0, 35) for i in x]

plt.figure(figsize=(10, 5)) # 设置图形大小
plt.plot(x, y_bj, label='北京') # 绘制北京的折线图
plt.plot(x, y_xa, label='西安') # 绘制西安的折线图

plt.title("气温变化折线图") # 设置标题
plt.xlabel("时间") # 设置x轴标签
plt.ylabel("气温") # 设置y轴标签
plt.xticks(x) # 设置x轴刻度
y_ticks = [i for i in range(0, 35, 2)] # 设置y轴刻度范围为0-35，步长为2
plt.yticks(y_ticks) # 设置y轴刻度
plt.legend(loc='upper left') # 显示图例
plt.grid(linestyle='--', alpha=0.5) # 显示网格

plt.show() # 显示图形
```
